```python
# This computes the size of the conjugacy class corresponding to an input partition
def classSize(p):
    p=list(p)
    n=sum(p)
    r=max(p)
    return factorial(n)//product(p)//product([factorial(p.count(i)) for i in range(1,r+1)])
```

For each $n$, we want to find a distribution of latin square completions such that a non-


Objective function $u\cdot v$, where we know what $v$ is. So let $u=(x_1,x_2,\dots,x_{P(n)})$ subject to the constraint that $\sum x_i = Number of latin squares$ and $x_i$ are non-negative integers.


Linear program:
- Objective function is the eigenvalue. We are trying to minimize.
- Variables represent the latin square counts by derangement.
- Constraints are:
    - Variable are in $\mathbb{R}$ for computation speed.
    - Total number of latin squares are correct.
    - The pairwise ratios of each completion count are at most the max_ratio.

```python
max_ratio = 2
for n in range(4,6):
    print(f"n={n}")
    program = MixedIntegerLinearProgram(maximization=False)
    P=Partitions(n).list()
    scaled_diagonal_entry=standardized_latin_squares[n]*(n-1)
    print(scaled_diagonal_entry)
    for mu in P:
        if (1 not in mu):
            print(mu,end=" ")
    print("\n")
    for lam in P: 
        fixed_calcs = [(classSize(mu)*charTable(lam,mu)/charTable(lam,[1]*n),classSize(mu)) for mu in P if 1 not in mu]
        number_of_partitions = len(fixed_calcs)
        x = program.new_variable(real=True,nonnegative=True)
        program.add_constraint(program.sum(x[i]*fixed_calcs[i][1] for i in range(number_of_partitions)) == scaled_diagonal_entry)
        for i in range(number_of_partitions):
            for j in range(number_of_partitions):
                if (i!=j):
                    program.add_constraint(x[i] <= max_ratio*x[j])
            
        program.set_objective(standardized_latin_squares[n]+program.sum(x[i]*fixed_calcs[i][0] for i in range(number_of_partitions)))
        print(f"{lam}", end=" "*(35-len(str(lam))))
        print("Least possible eigenvalue:",end=" ")
        print(round(program.solve(),2), end=" ")
        print({i: round(v,1) for i, v in program.get_values(x).items()},end=" ")
        print(f"Sum of latin squares: {sum([list(program.get_values(x).values())[i]*fixed_calcs[i][1] for i in range(number_of_partitions)])}")
    print("-"*80)
```
